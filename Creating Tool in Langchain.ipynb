```python
from langchain_core.tools import tool

@tool
def multiply(a,b):
  """Multiply two numbers"""
  return a*b

print(multiply.name)
print(multiply.description)
print(multiply.args)
```

```python
from pydantic import BaseModel, Field
from langchain_core.tools import StructuredTool

class CalculatorInput(BaseModel):
  a: float = Field(description="The first number")
  b: float = Field(description="The second number")

def multiply (a: float, b: float) -> float:
  """Multiply two numbers"""
  return a*b

multiply = StructuredTool.from_function(
    func=multiply,
    name="multiply",
    description="useful for when you need to multiply two numbers",
    args_schema=CalculatorInput,
    return_direct=True,
)
print(multiply.name)
print(multiply.description)
print(multiply.args)
```

```python
multiply.invoke({"a":2,"b":3})
```

```python
multiply.invoke({"a":2,"b":'abc''})
```

```python
from markitdown import MarkItDown
from langchain_community.tools.tavily_search import TavilySearchResults
from tqdm import tqdm
import requests

tavily_tool=TavilySearchResults(max_results=5,
                                search_depth='advanced',
                                include_answer=False,
                                include_raw_content=True)
md=MarkItDown()

@tool
def search_web_extract_info(query: str) -> list:
  """Search the web for a query extracts useful information fromt he search links"""
  results = tavily_tool.invoke(query)
  docs = []
  for result in tqdm(results):
    try:
      extracted_infor = md.converted(result['url'])
      text_title = extracted_infor.title.strip()
      text_content = extracted_infor.text_content.strip()
      docs.append(text_title + '\n'+ text_content)
    except:
      print("Extraction blocked fro url: ', result['url']")
      pass
  return docs
```

```python
docs = search_web_extract_info('OpenAI GPT-4o')
```

```python
from IPython.display import display, Markdown

if docs:
    display(Markdown(docs[0]))
else:
    display(Markdown("The 'docs' list is empty. Please ensure the `search_web_extract_info` function executed successfully and populated the list."))
```
